# Experiment 11: K-Means and Hierarchical Clustering
Use the elbow method to choose K and compare two clustering methods.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering

CSV_FILE = "Mall_Customers.csv"
df = pd.read_csv(CSV_FILE)
features = ["Annual Income (k$)", "Spending Score (1-100)"]
X = df[features].dropna()
X_scaled = StandardScaler().fit_transform(X)
```

```python
inertia = []
for k in range(1, 11):
    inertia.append(KMeans(n_clusters=k, n_init=10, random_state=42).fit(X_scaled).inertia_)
plt.plot(range(1, 11), inertia, marker="o")
plt.xlabel("Number of clusters")
plt.ylabel("WCSS")
plt.title("Elbow method")
plt.show()
```

```python
K = 5  # Change after observing the elbow plot
kmeans_labels = KMeans(n_clusters=K, n_init=10, random_state=42).fit_predict(X_scaled)
hierarchical_labels = AgglomerativeClustering(n_clusters=K).fit_predict(X_scaled)
print("K-Means labels:", kmeans_labels)
print("Hierarchical labels:", hierarchical_labels)
```
