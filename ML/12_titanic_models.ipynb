# Experiment 12: Titanic Survival Prediction
Preprocess passenger data and compare Logistic Regression, Random Forest, and SVM.

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

CSV_FILE = "train.csv"
df = pd.read_csv(CSV_FILE)
X = df.drop(columns=["Survived", "PassengerId", "Name", "Ticket", "Cabin"], errors="ignore")
y = df["Survived"]
numeric = X.select_dtypes("number").columns
categorical = X.select_dtypes(exclude="number").columns
preprocess = ColumnTransformer([("numbers", Pipeline([("fill", SimpleImputer(strategy="median")), ("scale", StandardScaler())]), numeric), ("categories", Pipeline([("fill", SimpleImputer(strategy="most_frequent")), ("onehot", OneHotEncoder(handle_unknown="ignore"))]), categorical)])
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42, stratify=y)
```

```python
models = {"Logistic Regression": LogisticRegression(max_iter=1000), "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42), "SVM": SVC()}
for name, classifier in models.items():
    model = Pipeline([("preprocess", preprocess), ("classifier", classifier)])
    model.fit(X_train, y_train)
    print(name, "accuracy:", round(accuracy_score(y_test, model.predict(X_test)), 3))
```
